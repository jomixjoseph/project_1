```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df_6,
    x="delivery_delay",
    y="total_order_value"
)

plt.title("Order Value vs Delivery Delay")
plt.xlabel("Delivery Delay (Days)")
plt.ylabel("Total Order Value")

plt.show()
```

```python
#Trend analysis
```

```python
print(df_6["order_purchase_Date"].head())
print(df_6["order_purchase_Date"].dtype)
```

```python
df_6["order_purchase_Date"] = pd.to_datetime(
    df_6["order_purchase_Date"]
)
```

```python
monthly_sales = df_6.groupby(
    df_6["order_purchase_Date"].dt.to_period("M")
)["total_order_value"].sum().reset_index()

monthly_sales["order_purchase_Date"] = monthly_sales[
    "order_purchase_Date"
].astype(str)

monthly_sales.head()
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    monthly_sales["order_purchase_Date"],
    monthly_sales["total_order_value"]
)

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Total Order Value")

plt.xticks(rotation=45)
plt.show()
```

```python
monthly_orders = df_6.groupby(
    df_6["order_purchase_Date"].dt.to_period("M")
)["order_id"].count().reset_index()

monthly_orders["order_purchase_Date"] = monthly_orders[
    "order_purchase_Date"
].astype(str)

monthly_orders.head()
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    monthly_orders["order_purchase_Date"],
    monthly_orders["order_id"]
)

plt.title("Monthly Order Trend")
plt.xlabel("Month")
plt.ylabel("Number of Orders")

plt.xticks(rotation=45)
plt.show()
```

```python
monthly_avg_order = df_6.groupby(
    df_6["order_purchase_Date"].dt.to_period("M")
)["total_order_value"].mean().reset_index()

monthly_avg_order["order_purchase_Date"] = monthly_avg_order[
    "order_purchase_Date"
].astype(str)

monthly_avg_order.head()
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    monthly_avg_order["order_purchase_Date"],
    monthly_avg_order["total_order_value"]
)

plt.title("Monthly Average Order Value Trend")
plt.xlabel("Month")
plt.ylabel("Average Order Value")

plt.xticks(rotation=45)
plt.show()
```
