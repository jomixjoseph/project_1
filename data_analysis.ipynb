```python
import numpy as np
import pandas as pd
```

```python
df_3 = pd.read_csv("cleaned/olist_order_items_dataset_cleaned.csv")
```

```python
df_3.head(10)
```

```python
Q1 = df_3["price"].quantile(0.25)
Q3 = df_3["price"].quantile(0.75)

IQR = Q3 - Q1
lower_fence = Q1 - 1.5 * IQR
upper_fence = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Fence:", lower_fence)
print("Upper Fence:", upper_fence)
```

```python
iqr_outliers = df_3[
    (df_3["price"] < lower_fence) |
    (df_3["price"] > upper_fence)
]

iqr_outliers[
    ["order_id", "price"]
]
```
